```python
#data3['host_location']       
#Cuarto método de sustitución de valores nulos
#Sustituir valores nulos por un string en  concreto 
data3["host_location"] =data3["host_location"].fillna("Ottawa, Canada") 
```

```python
#Tercer método de sustitución de valores nulos
#Sustituir valores nulos por un valor numérico en  concreto 
data3["first_review"]= data3["first_review"].fillna(10)
data3["last_review"]= data3["last_review"].fillna(10)  
```

```python
#Primer método de sustitución de valores nulos
#Sustituir valores nulos con promedio o media
data3["review_scores_rating"]=data3["review_scores_rating"].fillna(round(data3["review_scores_rating"].mean(),1))
data3["review_scores_accuracy"]=data3["review_scores_accuracy"].fillna(round(data3["review_scores_accuracy"].mean(),1))
data3["review_scores_checkin"]=data3["review_scores_checkin"].fillna(round(data3["review_scores_checkin"].mean(),1))
data3["review_scores_communication"]=data3["review_scores_communication"].fillna(round(data3["review_scores_communication"].mean(),1))
data3["review_scores_location"]=data3["review_scores_location"].fillna(round(data3["review_scores_location"].mean(),1))
data3["review_scores_value"]=data3["review_scores_value"].fillna(round(data3["review_scores_value"].mean(),1))
data3["reviews_per_month"]=data3["reviews_per_month"].fillna(round(data3["reviews_per_month"].mean(),1))
```

```python
#Primer método de sustitución de valores nulos
#Sustituir valores nulos con promedio o media
data3["bathrooms"]=data3["bathrooms"].fillna(round(data3["bathrooms"].mean(),1))
data3["bedrooms"]=data3["bedrooms"].fillna(round(data3["bedrooms"].mean(),1))
data3["beds"]=data3["beds"].fillna(round(data3["beds"].mean(),1))
```

```python
#Primer método de sustitución de valores nulos
#Sustituir valores nulos con promedio o media
data3["price_quote_total_price"]=data3["price_quote_total_price"].fillna(round(data3["price_quote_total_price"].mean(),1))
data3["price_quote_price_per_night"]=data3["price_quote_price_per_night"].fillna(round(data3["price_quote_price_per_night"].mean(),1))
```

```python
#Cuarto método de sustitución de valores nulos
#Sustituir valores nulos por un string en  concreto
data3["price_quote_checkout_date"] =data3["price_quote_checkout_date"].fillna("0000-00-00") 
```

```python
#Cuarto método de sustitución de valores nulos
#Sustituir valores nulos por un string en  concreto
##data3["price_quote_checkin_date"] =data3["price_quote_checkin_date"].fillna(10) >> la super regue
data3["price_quote_checkin_date"] = data3["price_quote_checkin_date"].replace(['00:00:00'], "0000-00-00")
```

```python
# Convertir price a número
data3["price"] = data3["price"].str.replace("$", "").str.replace(",", "").astype(float)

#Primer método de sustitución de valores nulos
#Sustituir valores nulos con promedio o media
data3["price"] = data3["price"].fillna(round(data3["price"].mean(), 1))
```

```python
#Primer método de sustitución de valores nulos
#Sustituir valores nulos con promedio o media
data3["minimum_nights"] = data3["minimum_nights"].fillna(round(data3["minimum_nights"].mean(), 1))
data3["maximum_nights"] = data3["maximum_nights"].fillna(round(data3["maximum_nights"].mean(), 1))
```

```python
#has_availability
#Cuarto método de sustitución de valores nulos
#Sustituir valores nulos por un string en  concreto
data3["has_availability"] =data3["has_availability"].fillna("f") 
```
